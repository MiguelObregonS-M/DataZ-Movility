```python
# ============================================
# 01_exploracion.ipynb — Exploración inicial
# Proyecto: DataZ Movility (Fase A — Bizi)
# Autor: Miguel
# ============================================

import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
import seaborn as sns
import json
from pathlib import Path

# Ajustes visuales
plt.style.use("seaborn-v0_8")
sns.set_palette("viridis")

# Rutas del proyecto
DATA_RAW = Path("../data/raw")
DATA_PROCESSED = Path("../data/processed")

print("Entorno cargado correctamente.")
```

```python
# Cargar el archivo bizi.geojson
bizi_path = DATA_RAW / "bizi.geojson"

try:
    gdf_bizi = gpd.read_file(bizi_path)
    print("Dataset Bizi cargado correctamente.")
except Exception as e:
    print("Error al cargar el dataset:", e)

gdf_bizi.head()
```

```python
# Información general del dataset
gdf_bizi.info()
```

```python
# Vista rápida de estadísticas numéricas
gdf_bizi.describe(include="all")
```

```python
print("Columnas disponibles:")
list(gdf_bizi.columns)
```

```python
# Mostrar 5 registros con columnas relevantes
gdf_bizi[[
    "id",
    "title",
    "bicisDisponibles",
    "anclajesDisponibles",
    "estado",
    "geometry"
]].head()

```

```python
# Valores nulos por columna
gdf_bizi.isnull().sum()
```

```python
# Comprobación de duplicados por ID
gdf_bizi["id"].duplicated().sum()
```

```python
# Rango de bicis disponibles
# Rango de anclajes libres
print("Rango de bicis:", gdf_bizi["bicisDisponibles"].min(), "-", gdf_bizi["bicisDisponibles"].max())
print("Rango de anclajes:", gdf_bizi["anclajesDisponibles"].min(), "-", gdf_bizi["anclajesDisponibles"].max())
```

```python
# Detectar valores negativos o inconsistentes
anomalos = gdf_bizi[
    (gdf_bizi["bicisDisponibles"] < 0) |
    (gdf_bizi["anclajesDisponibles"] < 0)
]
anomalos

```

```python
#historgramas
plt.figure(figsize=(10,5))
sns.histplot(gdf_bizi["bicisDisponibles"], bins=20, kde=True)
plt.title("Distribución de bicis disponibles")
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.histplot(gdf_bizi["anclajesDisponibles"], bins=20, kde=True)
plt.title("Distribución de anclajes disponibles")
plt.show()

```

```python
# Mapa simple usando geopandas
gdf_bizi.plot(figsize=(8,8), markersize=10, color="blue")
plt.title("Mapa preliminar de estaciones Bizi")
plt.show()
```

```python
# Crear columna de ocupación (bikes / capacidad total)
gdf_bizi["capacity"] = gdf_bizi["bicisDisponibles"] + gdf_bizi["anclajesDisponibles"]
gdf_bizi["ratio_ocupacion"] = gdf_bizi["bicisDisponibles"] / gdf_bizi["capacity"]

gdf_bizi[[
    "id",
    "title",
    "bicisDisponibles",
    "anclajesDisponibles",
    "capacity",
    "ratio_ocupacion"
]].head()

```

```python
# Guardar dataset preliminar para la fase de limpieza
output_path = DATA_PROCESSED / "bizi_preliminar.geojson"
gdf_bizi.to_file(output_path, driver="GeoJSON")

print("Dataset preliminar guardado en:", output_path)
```
